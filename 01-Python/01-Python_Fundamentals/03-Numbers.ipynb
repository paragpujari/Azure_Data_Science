##### Basic Numerical operations

| Operator | Meaning        |   Example | Result |
| -------- | -------------- | --------: | -----: |
| `+`      | Addition       |  `10 + 5` |   `15` |
| `-`      | Subtraction    |  `10 - 5` |    `5` |
| `*`      | Multiplication |  `10 * 5` |   `50` |
| `/`      | Division       |  `10 / 5` |  `2.0` |
| `//`     | Floor division | `10 // 3` |    `3` |
| `%`      | Remainder      |  `10 % 3` |    `1` |
| `**`     | Power          |  `2 ** 3` |    `8` |


```python
x = 10
y = 3

##### Mathematical operations

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x % y)
print(x // y)
print(x ** y)
```

22. Check Whether a Number is Palindrome

```python
##### enter the number
num = int(input('Enter the number'))

##### Check whether the number is palindrome or not
rev = 0
x = num
while(x != 0):
    d = x%10
    rev = (rev*10)+d
    x = x//10
if(rev==num):
    print(f'{num} is palindrome')
else:
    print(f'{num} is not palindrome')
```

26.  Find the smallest digit in a positive integer

```python
##### enter the number
num = int(input('Enter the number'))

##### Find the smallest digit in a positive integer
smallest = 1
while(num != 0):
    d = num%10
    if(d <= smallest):
        smallest = d
    num = num//10
print(f'Smallest digit in the number is {smallest}')
```

27. Given a number, count how many digits are even and how many are odd

```python
##### enter the number
num = int(input('Enter the number'))

##### Count the number of even and odd digits
even_dig = 0
odd_dig  = 0

while(num != 0):
    d = num%10
    if(d%2==0):
        even_dig = even_dig + 1
    else:
        odd_dig = odd_dig  + 1
    num = num//10

print(f'Number of even digits is {even_dig}')
print(f'Number of odd  digits is {odd_dig}')
```

28.  Check Armstrong Number

```python
##### enter the number
num = int(input('Enter the number'))


##### Check if the number is Armstrong or not
sum=0
x=num 

while(x != 0):
    d = x%10
    sum = sum+(d**3)
    x = x//10

if(sum==num):
    print(f'{num} is armstrong')
else:
    print(f'{num} is not armstrong')
```

29.  Find GCD of Two Numbers

```python
##### enter the two numbers
a = int(input('Enter the first number'))
b = int(input('Enter the second number'))

while(b != 0):
    rem = a%b
    a   = b
    b   = rem 
print(a)
```
