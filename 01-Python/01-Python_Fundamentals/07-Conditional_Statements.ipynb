##### Conditional Statements

1.  These are the selection statements that is used to make certain selections based on the conditions.

Types of Conditional Statements

1.  if statements


if condition:

    statement

2. if ... else

Use else when there are two possible outcomes


if condition:

    statement1
else:

    statement2


3. if ... elif ... else

Use elif when there are multiple conditions


if condition1:

    statement1

elif condition2:

    statement2

else:

    statement3

### 4. Comparison Conditions

These are commonly used with if:

| Operator | Meaning               |
| -------- | --------------------- |
| `>`      | greater than          |
| `<`      | less than             |
| `>=`     | greater than or equal |
| `<=`     | less than or equal    |
| `==`     | equal                 |
| `!=`     | not equal             |


#### 5. and, or, not inside conditions

These are used only to combine conditions.

and

Both conditions must be true


or

At least one condition must be true


not

Reverses the condition


#### 6. Nested if

An if can exist inside another if

age = 25

citizen = True

if age >= 18:

    if citizen:

        print("Eligible")

30.  Production incident severity

A production system classifies an incident using impact and downtime.

Rules:

Impact >= 90 AND downtime >= 60 → Critical

Impact >= 70 AND downtime >= 30 → High

Impact >= 40 → Medium

Otherwise → Low

```python
impact = 75
downtime = 40

if((impact >= 90) and (downtime >= 60)):
    print("Critical")
elif((impact >= 70) and (downtime >= 30)):
    print("High")
elif(impact >= 40):
    print("Medium")
else:
    print("Low")
```
