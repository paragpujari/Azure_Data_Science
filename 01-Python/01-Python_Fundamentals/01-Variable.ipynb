##### Properties of Python

1.  Python is dynamically typed.

2.  We do not need to explicitly define the data type of the variable. Its data type is automatically defined when we assign the value to the variable.

3.  The value of the variable changes and varies during the runtime execution of the program.

```python
x = 100     ##### int
print(x)
print(type(x))


x = 'Anish'  ##### string
print(x)
print(type(x))


x = True      ##### Boolean
print(x)
print(type(x))
```

6. It should not used any special character like ($,%,&) etc.

##### Intermediate Level

1. Employee Salary Calculation

You have:

basic_salary = 50000

hra_percentage = 20

da_percentage = 10

tax_percentage = 5

Calculate:

HRA

DA

Gross salary

Tax

Net salary

Use:

HRA = basic_salary × HRA%

DA = basic_salary × DA%

Gross Salary = basic + HRA + DA

Tax = Gross Salary × Tax%

Net Salary = Gross Salary - Tax

Expected output:

HRA: 10000

DA: 5000

Gross Salary: 65000

Tax: 3250

Net Salary: 61750

```python
basic_salary = 50000

hra_percentage = 20

da_percentage = 10

tax_percentage = 5


##### Calculate the HRA

hra = (basic_salary * hra_percentage)/100.0
print(f'HRA salary is {hra}')

##### Calculate the DA
da = (basic_salary * da_percentage)/100.0
print(f'DA salary is {da}')

##### Calculate the Gross salary
gross_salary = basic_salary + hra + da
print(f'Gross salary is {gross_salary}')

##### Calculate the tax
tax = (gross_salary * tax_percentage)/100.0
print(f'Tax salary is {tax}')

##### Calculate the net salary
net_salary = gross_salary - tax
print(f'Net salary is {net_salary}')
```

2.  Problem

Given an integer n, return a list of strings from 1 to n.

For each number:

If divisible by 3 and 5 → "FizzBuzz"

If divisible by 3 → "Fizz"

If divisible by 5 → "Buzz"

Otherwise → the number itself as a string

```python
##### enter the number
result = []
num = int(input('Enter the number'))

for i in range(1,(num+1)):
    if((i%3)==0 and (i%5)==0):
        result.append("FizzBuzz")
    elif((i%3)==0):
        result.append("Fizz")
    elif((i%5)==0):
        result.append("Buzz")
    else:
        result.append(str(i))

print(result)
```

Question 1 — ETL Pipeline Success Analysis

You are monitoring a daily data pipeline.

The pipeline processed:

total_records = 125000

successful_records = 121500

failed_records = 3500

Business Problem

The data engineering team wants to know:

What percentage of records were successfully processed?

What percentage failed?

Does successful_records + failed_records equal total_records?

If the acceptable failure rate is below 2%, should the pipeline be considered acceptable?

How many additional records need to succeed to bring the failure rate below 2%?

Your Task

Write Python code using variables and basic calculations.

Expected analytical thinking

First calculate:

Failure Rate = Failed Records / Total Records × 100

Then compare the result with:

2%

Important: Don't immediately write code. First identify:

Input → Calculation → Validation → Business Decision

Expected output format

Total Records: 125000

Successful Records: 121500

Failed Records: 3500

Success Rate: XX%

Failure Rate: XX%

Data Validation: True

Required Additional Successful Records: XXXX

```python
total_records = 125000

successful_records = 121500

failed_records = 3500


##### 1. What percentage of records were successfully processed?
success_rate = (successful_records * 100.0)/total_records
print(f'Success Percentage is {success_percentage}')


##### 2. What percentage failed?
failure_rate = (failed_records * 100.0)/total_records
print(f'Success Percentage is {failure_rate}')

##### 3. Does successful_records + failed_records equal total_records?
if((successful_records + failed_records) == total_records):
    print(f'successful_records + failed_records equals to total_records')
else:
    print(f'successful_records + failed_records not equals to total_records')

##### 4. If the acceptable failure rate is below 2%, should the pipeline be considered acceptable?
acceptable_failure_rate = 2
actual_failure_rate = 2.8

if(actual_failure_rate < acceptable_failure_rate):
    print(f'Pipeline is considered to be acceptable')
else:
    print(f'Pipeline is not considered to be acceptable')

##### 5. How many additional records need to succeed to bring the failure rate below 2%?
failure_records = (total_records * 2)/100.0
print(failure_records)

additional_records = failed_records - failure_records
print(additional_records)
```
