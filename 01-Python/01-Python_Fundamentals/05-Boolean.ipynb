##### Boolean

1.  A Boolean has two values. One is True and the other is False.

2.  It has one data type that is bool.

```python
##### Boolean Value
a = True
b = False

print(a)
print(b)


print(type(a))
print(type(b))
```

##### Creating Boolean Values

1.  We can directly create boolean variables and assign boolean values to it

```python
is_student = False
is_employed = True

print(is_student)
print(type(is_student))
print('----------------------------------------------------------------------------')
print(is_employed)
print(type(is_employed))
```

```python
is_logged_in = True
print(is_logged_in)
print(type(is_logged_in))
print('----------------------------------------------------------------------------')
is_active = True
print(is_active)
print(type(is_active))
print('----------------------------------------------------------------------------')
is_verified = False
print(is_verified)
print(type(is_verified))
print('----------------------------------------------------------------------------')
is_admin = False
print(is_admin)
print(type(is_admin))
print('----------------------------------------------------------------------------')
```
