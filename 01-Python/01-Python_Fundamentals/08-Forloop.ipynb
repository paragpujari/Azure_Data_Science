##### For Loop

A Python for loop is used when we want to repeat a block of code for every item in a sequence or range


for variable in sequence:
    statement


1.  range(start, stop,step) ---------> It helps in printing the series of all the numbers from starting till (end-1) based on the step interval.

```python
for i in range(1,11,1):
    print(i)
```

##### Intermediate Level Interview Questions

11. Calculate the sum from 1 to 10

```python
s=0
for i in range(1,11):
    s=s+i
print(f'Sum of all the numbers from 1...10 is {s}')
```

12. Calculate the sum of even numbers from 1 to 20

```python
se=0
for i in range(1,21):
    if((i%2)==0):
        se=se+i
print(f'Sum of all the even numbers from 1...20 is {se}')       
```

13. Calculate the sum of odd numbers from 1 to 20

```python
so=0
for i in range(1,21):
    if((i%2)!=0):
        so=so+i
print(f'Sum of all the odd numbers from 1...20 is {so}') 
```

14.  Calculate the product of numbers from 1 to 5

```python
pro=1
for i in range(1,6):
    pro = pro * i
print(f'Product of all the numbers from 1...5 is {pro}')
```

15. Print the multiplication table of 5

```python
for i in range(1,11):
    print(f'5 * {i} = {5*i}')
```

16.  Print the multiplication table of 7 in proper format

```python
for i in range(1,11):
    print(f'7 * {i} = {7*i}')
```

17.  Count how many numbers exist from 1 to 100

```python
c=0
for i in range(1,101):
    c=c+1
print(f'Total numbers between 1...100 is {c}' )
```

18.  Find the sum of multiples of 3 from 1 to 30

```python
s=0
for i in range(1,31):
    if((i%3)==0):
        s=s+i
print(f'Sum of all the multiplies of 3 from 1 to 30 is {s}')
```

19.  Print numbers in reverse with a step of 2

```python
for i in range(20,0,-2):
    print(i)
```

20. Calculate the sum of squares from 1 to 5

```python
s=0
for i in range(1,6):
    s = s + (i**2)
print(f'Sum of all the squares from 1...5 is {s}')
```
