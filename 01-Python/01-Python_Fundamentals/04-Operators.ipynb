##### Operators

1.  Operators can be defined as the special symbols that is used to perform certain mathematical operations in Python.

##### Types of Operators

There are seven type of Operators. They are as follows:---

(a.)    Arithmetic Operators

(b.)    Relational Operators

(c.)    Logical Operators

(d.)    Bitwise Operators

(e.)    Assignment Operators

(f.)    Membership Operators

(g.)    Identity Operators

##### (a.)    Arithmetic Operators

```python
a = 4
b = 2

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

##### (b.)    Relational Operators

```python
a = 4
b = 2

print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
print(a == b)
print(a != b)
```

##### (c.) Logical Operators

```python
a=0
b=1

print(a and b)
print(a or  b)
print(not b)
print(not a)
```

##### (d.) Bitwise Operators

```python
a=2
b=3

print(a&b)    ##### Bitwise AND
print(a|b)    ##### Bitwise OR
print(~a)     ##### Bitwise NOT
print(a^b)    ##### Bitwise XOR
print(4>>2)   ##### Bitwise Left shift
print(5 << 2) ##### Bitwise Right shift
```

##### (e.) Assignment Operators

```python
a = 10
a+=2
print(a)


b = 10
b-=2
print(b)

c = 10
c*=2
print(c)

d = 10
d/=2
print(d)

e = 10
e//=2
print(e)

f = 10
f%=2
print(f)

g = 10
g**=2
print(g)
```

##### (f.)    Membership Operators

```python
lists = [1,2,3,4,5]

print(6 in lists)

print(10 not in lists)
```

```python
print(1 in lists)

print(5 not in lists)
```

##### (g.)    Identity Operators

```python
x = 10

y = 10

print(id(x))

print(id(y))

print(x is y)   ###### It determines whether both variables have the same memory address or not
```

```python
x = [1,2,3]

y = x.copy()   ##### shallow copy

y.append(4)

print(x)
print(y)

print(id(x))

print(id(y))

print(x is y)
```
