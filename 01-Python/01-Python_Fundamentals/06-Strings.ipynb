##### Functions in Strings

1.  lower() ------------>  It is used to convert the string into lower case.

```python
text = "PYTHON"

##### Convert the text into lower case
res = text.lower()
print(res)
```

2. upper() ----------> It is used to convert the string into upper case.

```python
text = "python"

##### Convert the text into upper case
res = text.upper()
print(res)
```

3.  strip()

Removes spaces from the beginning and end

```python
text = "    Python  "

##### Remove the blankspaces from both sides of the string
res = text.strip()

print(res)
```

4.  replace()

It is used to replace one portion of the string with another string.

```python
text = "I like JAVA"

##### Replace JAVA with Python from the text
res = text.replace("JAVA","Python")
print(res)
```

5. find()

It is used to find out the index position of the character from the string.

```python
text = "Python"

##### Find out the index position of the character from the string
res = text.index('y')
print(res)
```

6. count()

It is used to count the occurances of the particular character in the string

```python
text = "banana"

##### Count the occurances of 'a' from the text

res = text.count('a')
print(res)
```

7. startswith()

It is used to check if the string starts with bthe particular occurance

```python
text = "Python"

##### Check if the text starts with "Py"
res = text.startswith("Py")
print(res)
```

8. endswith()

It is used to check if the string ends with the particular occurance

```python
text = "Python"

##### Check if the text ends with "on"
res = text.endswith("on")
print(res)
```

2. alnum()

```python
text = "PyThOn"

upper=0
lower=0
for x in text:
    if(x.isupper()):
        upper=upper+1
    else:
        lower = lower+1
print(f'Number of uppercase characters is {upper}')
print(f'Number of lowercase characters is {lower}')
```
