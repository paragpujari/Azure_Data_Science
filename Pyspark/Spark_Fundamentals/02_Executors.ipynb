```python
##### import all the spark libraries
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum,col
```

```python
##### Creation of Spark Session --------> The driver creates the spark session
session = SparkSession.builder.appName('Builder').getOrCreate()

##### get the version of spark session
print(session.version)
```

```python
##### Creation of dataframe
data = [
    (101, "Rahul", "IT", 50000),
    (102, "Priya", "HR", 45000),
    (103, "Amit", "IT", 60000),
    (104, "Neha", "Finance", 55000),
    (105, "Ravi", "IT", 65000)
]

columns = ['employee_id','employee_name','department','salary']


##### Construct the dataframe
df = session.createDataFrame(data,columns)

df.show()
```

```python
##### Perform the actual execution i.e. aggrgration of all the salaries of the employees
df.agg(sum(col('salary')).alias('Total_salary')).show()
```

#### **OBSERVATIONS:**

1.  Here we have considered all the values in the salary column.

2.  Then we have performed the actual execution of the dataframe(i.e aggrgation of all the salaries of all the employees)

3. Then the total salary(sum of all the employees) is then sent to the driver program.
