##### **Spark Session**

1.  A Spark Session can be defined as the entry point through which the particular user can access all the functionalities of Pyspark.

```python
##### import all the libraries of spark session
from pyspark.sql           import SparkSession
from pyspark.sql.functions import col,sum
```

```python
##### Create the session for spark
session = SparkSession.builder.appName('SparkSession').getOrCreate()

##### Get the version of spark
print(session.version)
```

```python
##### Create the dataframe from the data using SparkSession
data = [
    (1, "Parag", 60000),
    (2, "Rahul", 70000),
    (3, "Amit", 80000),
    (4, "Priya", 90000)
]

columns = ["employee_id", "employee_name","salary"]

##### Create the dataframe
df = session.createDataFrame(data,columns)

##### Display the dataframe
df.show()
```

```python
##### Display the schema of the dataframe
df.printSchema()
```

```python
##### Select the columns
df.select("employee_id","employee_name").show()
```

```python
##### Filter all the employees whose salary > 70000
df.filter(df.salary > 70000).show()
```

##### **Spark Session also allows us to use Spark SQL**

```python
##### Creation of Temporary table using Spark SQL
df.createOrReplaceTempView('employees')
```

```python
##### Spark Session interacts with Spark SQL
result = session.sql("""
SELECT employee_name,salary FROM employees WHERE salary > 70000
""")
```

```python
result.show()
```
