```python
##### import all the libraries for spark session
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum,col
```

```python
##### Create the Spark Session
session = SparkSession.builder.appName('Workernode').getOrCreate()
##### Get the version of spark session
print(session.version)
```

```python
##### Create the dataframe
data = [
    ("A", 50000),
    ("B", 60000),
    ("C", 70000),
    ("D", 80000),
    ("E", 90000),
    ("F", 100000)
]

columns = ["employee","salary"]

df = session.createDataFrame(data,columns)

df.show()
```

```python
##### Get the sum of salaries for all the employees
ans=df.agg(sum(col('salary')).alias('Total_salary'))
```

```python
ans.show()
```

##### **OBSERVATIONS:**

1.  The dataset is divided into partitions.

2.  Then all the partitions are then sent to the worker node.

3.  The worker node sends the data into the executor fo data processing.

4.  The executor executes the data and gives us the results.

5.  Then all the results are collected from each and every executor and the combined result is sent to the driver program.
