**1. Simple PySpark program that calculates the average salary by department**

```python
##### import all the necessary libraries of pyspark
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,avg
```

```python
##### Create a spark session
session = SparkSession.builder.appName('ClusterManager').getOrCreate()

##### Get the version of spark session
print(session.version)
```

```python
##### Create the dataframe
data = [
    (1, "Rahul", "IT", 60000),
    (2, "Amit", "IT", 80000),
    (3, "Neha", "HR", 50000),
    (4, "Priya", "HR", 70000)
]

columns = ["id", "name", "department", "salary"]

df = session.createDataFrame(data,columns)

df.show()
```

```python
##### Get the Average salary in department wise
df.groupby(col('department')).agg(avg(col('salary')).alias('Average_salary')).show()
```

#### **OBSERVATIONS:**

1.  Here Cluster Manager plays an important role in allocating all the resources in the cluster that is required to execute the spark application.
