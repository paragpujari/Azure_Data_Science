```python
##### import all the libraries for pyspark
from pyspark.sql           import SparkSession
from pyspark.sql.functions import sum

##### Create a spark session ----------> The driver creates the spark session
session = SparkSession.builder.appName('Driver').getOrCreate()

##### get the version of spark session
print(session.version)
```

#### **OBSERVATIONS:**

1.  The driver is the main component of spark application that creates the spark session and initializes the spark context.

```python
##### Create the dataset
data = [
    (101, "Rahul", "IT", 50000),
    (102, "Priya", "HR", 45000),
    (103, "Amit", "IT", 60000),
    (104, "Neha", "Finance", 55000),
    (105, "Ravi", "IT", 65000)
]

data

columns = ["employee_id", "emp_name","department","salary"]
```

```python
##### Construct the dataframe using spark session
df = session.createDataFrame(data,columns)
```
