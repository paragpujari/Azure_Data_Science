21.  The mean of 5 numbers is 20. One number is changed from 10 to 30. Find the new mean

```python
mean_old = 20
c = 5

sum_old = mean_old*c
print(sum_old)

##### One number changed from 10 ----> 30
old_value = 10
new_value = 30

value_diff = new_value - old_value
print(value_diff)

##### Caluclate the new mean
mean_new = sum_old + value_diff
print(mean_new)


new_mean = mean_new/c
print(f'New mean is {new_mean:.2f}')
```

30. Interview Challenge

A company's employee group has:

10 employees
Mean salary = ₹50,000

One employee with salary ₹40,000 leaves.

One new employee joins with salary ₹70,000.

Find the new mean salary.

```python
old_count = 10
old_mean = 50000

leaving_salary = 40000
joining_salary = 70000

old_total = old_count * old_mean

new_total = old_total - leaving_salary + joining_salary

new_count = old_count

new_mean = new_total / new_count

print("New Mean Salary:", new_mean)
```
